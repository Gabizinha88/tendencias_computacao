O objetivo deste projeto é desenvolver inicialmente uma calculadora básica em Python e, posteriormente, utilizar uma ferramenta de Inteligência Artificial para sugerir melhorias, refatorações e tratamentos de erros.

Desenvolvi inicialmente uma calculadora utilizando estruturas condicionais if e elif. A primeira versão realiza soma, subtração, multiplicação e divisão, mas ainda não possui tratamento de erros.

```python
print("CALCULADORA")

numero1 = float(input("Digite o primeiro número: "))
numero2 = float(input("Digite o segundo número: "))

print("1 - Somar")
print("2 - Subtrair")
print("3 - Multiplicar")
print("4 - Dividir")

opcao = input("Escolha uma operação: ")

if opcao == "1":
    resultado = numero1 + numero2
    print("Resultado:", resultado)

elif opcao == "2":
    resultado = numero1 - numero2
    print("Resultado:", resultado)

elif opcao == "3":
    resultado = numero1 * numero2
    print("Resultado:", resultado)

elif opcao == "4":
    resultado = numero1 / numero2
    print("Resultado:", resultado)

else:
    print("Opção inválida")
```

Durante os testes percebi que, ao tentar realizar uma divisão por zero, o programa apresentava um erro e interrompia sua execução. Entao sugeri a IA para melhorar meu codigo

Prompt enviado à IA:
“Estou desenvolvendo uma calculadora básica em Python. Ela realiza soma, subtração, multiplicação e divisão utilizando estruturas condicionais. Analise o código e sugira formas simples de organizá-lo melhor e tratar erros como divisão por zero, mantendo o código compreensível para um estudante iniciante.”

```python
def somar(a, b):
    return a + b

def subtrair(a, b):
    return a - b

def multiplicar(a, b):
    return a * b

def dividir(a, b):
    if b == 0:
        return "Erro: não é possível dividir por zero."
    return a / b


print("=== CALCULADORA ===")

numero1 = float(input("Digite o primeiro número: "))
numero2 = float(input("Digite o segundo número: "))

print("1 - Somar")
print("2 - Subtrair")
print("3 - Multiplicar")
print("4 - Dividir")

opcao = input("Escolha uma operação: ")

if opcao == "1":
    resultado = somar(numero1, numero2)

elif opcao == "2":
    resultado = subtrair(numero1, numero2)

elif opcao == "3":
    resultado = multiplicar(numero1, numero2)

elif opcao == "4":
    resultado = dividir(numero1, numero2)

else:
    resultado = "Opção inválida."

print("Resultado:", resultado)
```

A IA sugeriu separar cada operação em funções e adicionar uma verificação para divisão por zero. Considerei as alterações adequadas porque deixaram o código mais organizado e evitaram um erro que havia sido identificado durante os testes.

Segundo problema identificado: ao informar letras no lugar de números, o programa também encerra sua execução.

mandei um segundo prompt para refinar mais

“A calculadora já trata divisão por zero, mas ainda apresenta erro quando o usuário digita letras em vez de números. Como posso tratar esse problema de maneira simples usando Python?”

```python
def somar(a, b):
    return a + b

def subtrair(a, b):
    return a - b

def multiplicar(a, b):
    return a * b

def dividir(a, b):
    if b == 0:
        return "Erro: não é possível dividir por zero."
    return a / b


print("=== CALCULADORA ===")

try:
    numero1 = float(input("Digite o primeiro número: "))
    numero2 = float(input("Digite o segundo número: "))

    print("\n1 - Somar")
    print("2 - Subtrair")
    print("3 - Multiplicar")
    print("4 - Dividir")

    opcao = input("Escolha uma operação: ")

    if opcao == "1":
        resultado = somar(numero1, numero2)

    elif opcao == "2":
        resultado = subtrair(numero1, numero2)

    elif opcao == "3":
        resultado = multiplicar(numero1, numero2)

    elif opcao == "4":
        resultado = dividir(numero1, numero2)

    else:
        resultado = "Erro: opção inválida."

    print("Resultado:", resultado)

except ValueError:
    print("Erro: digite apenas números.")
```

A IA foi utilizada como ferramenta de apoio para sugerir refatorações e tratamentos de exceções. As sugestões foram analisadas antes de serem incorporadas ao código.
